## 1. Statistics

This section covers basic statistical concepts such as mean, median, mode, variance, and standard deviation.

```python
import numpy as np
import statistics

data = [2, 4, 6, 8, 10]

mean = np.mean(data)
median = np.median(data)
std = np.std(data)
variance = np.var(data)

print("Data:", data)
print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Standard Deviation:", std)
```

```python
data = [2, 3, 3, 4, 5, 3, 6]

mode = statistics.mode(data)

print("Data:", data)
print("Mode:", mode)
print("Range:", max(data) - min(data))
```

## 2. Probability

Probability measures how likely an event is to occur.

```python
# Probability of getting an even number when rolling a fair die

favorable_outcomes = 3
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
```

```python
# A bag contains 3 red balls and 2 blue balls

red_balls = 3
blue_balls = 2

total_balls = red_balls + blue_balls

probability_red = red_balls / total_balls

print("Probability of selecting a red ball:", probability_red)
```

```python
# P(A) = 0.7
# P(B) = 0.4
# A and B are independent

p_a = 0.7
p_b = 0.4

p_a_and_b = p_a * p_b

print("P(A and B):", p_a_and_b)
```

```python
# 10 students study Machine Learning
# 5 of them also study Python

python_and_ml = 5
ml_students = 10

conditional_probability = python_and_ml / ml_students

print("P(Python | ML):", conditional_probability)
```

## 3. Linear Algebra

Linear algebra is important in Machine Learning because data and model parameters are often represented using vectors and matrices.

```python
A = np.array([1, 2])
B = np.array([3, 4])

result = A + B

print("A + B =", result)
```

```python
A = np.array([2, 3])

result = 4 * A

print("4A =", result)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A + B

print("A + B:")
print(result)
```

```python
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

dot_product = np.dot(A, B)

print("Dot Product:", dot_product)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([5, 6])

result = np.dot(A, B)

print("Result:")
print(result)
```

## 4. Basic Calculus

Derivatives are important in Machine Learning because optimization algorithms such as Gradient Descent use gradients to update model parameters.

```python
# f(x) = x^2
# f'(x) = 2x

x = 3

derivative = 2 * x

print("Derivative at x = 3:", derivative)
```

```python
# f(x) = 3x^2 + 2x + 5
# f'(x) = 6x + 2

x = 2

derivative = 6 * x + 2

print("Derivative at x = 2:", derivative)
```

```python
# f(x) = x^3 - 4x
# f'(x) = 3x^2 - 4

x = 2

derivative = 3 * (x ** 2) - 4

print("Derivative at x = 2:", derivative)
```

## 5. Mathematics in Machine Learning

This example demonstrates how mathematical calculations are used to evaluate Machine Learning models.

```python
actual = np.array([10, 12, 8])
predicted = np.array([9, 11, 10])

errors = actual - predicted
squared_errors = errors ** 2

mse = np.mean(squared_errors)
rmse = np.sqrt(mse)

print("Actual values:", actual)
print("Predicted values:", predicted)
print("Errors:", errors)
print("Squared Errors:", squared_errors)
print("MSE:", mse)
print("RMSE:", rmse)
```

## 6. Short Questions and Answers

### Q1. What is the difference between mean and median?

The mean is calculated by adding all values and dividing by the number of values.
The median is the middle value after sorting the data.

### Q2. What is probability?

Probability measures how likely an event is to occur.

### Q3. Why are vectors and matrices important in Machine Learning?

They are used to represent data, features, weights, and mathematical operations inside Machine Learning models.

### Q4. Why are derivatives important in Machine Learning?

Derivatives help optimization algorithms determine how model parameters should change to reduce the loss function.

### Q5. What is RMSE?

RMSE stands for Root Mean Squared Error. It measures the difference between predicted and actual values, with larger errors receiving more weight.
