## 1. Variables and Data Types

This section demonstrates how to create variables and work with different Python data types.

```python
# Creating variables
name = "Arzaq"
age = 21
gpa = 3.7
is_student = True

# Printing values
print("Name:", name)
print("Age:", age)
print("GPA:", gpa)
print("Is Student:", is_student)

# Printing data types
print("\nData Types:")
print("Name:", type(name))
print("Age:", type(age))
print("GPA:", type(gpa))
print("Is Student:", type(is_student))
```

```python
# Basic calculations

length = 10
width = 5

area = length * width
perimeter = 2 * (length + width)

print("Area:", area)
print("Perimeter:", perimeter)

# Temperature conversion
celsius = 25
fahrenheit = (celsius * 9/5) + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

## 2. Lists

Lists are used to store multiple values in a single variable.

```python
# Creating a list of grades
grades = [85, 90, 78, 92, 88]

print("Grades:", grades)

# Accessing elements
print("First grade:", grades[0])
print("Last grade:", grades[-1])

# Adding an element
grades.append(95)
print("After adding a grade:", grades)

# Removing an element
grades.remove(78)
print("After removing a grade:", grades)

# Sorting
grades.sort()
print("Sorted grades:", grades)

# Calculations
average = sum(grades) / len(grades)

print("Average:", average)
print("Maximum:", max(grades))
print("Minimum:", min(grades))
```

## 3. Dictionaries

Dictionaries store data as key-value pairs.

```python
# Creating a student dictionary
student = {
    "name": "Arzaq",
    "age": 21,
    "major": "Artificial Intelligence",
    "gpa": 3.7
}

print("Student information:")
print(student)

# Accessing values
print("\nName:", student["name"])
print("Major:", student["major"])

# Adding a new key
student["university"] = "Minia National University"

# Updating a value
student["gpa"] = 3.8

print("\nUpdated student information:")
print(student)

# Looping through dictionary
print("\nStudent details:")
for key, value in student.items():
    print(key, ":", value)
```

## 4. Loops

Loops allow us to repeat a block of code multiple times.

```python
print("Numbers from 1 to 10:")

for i in range(1, 11):
    print(i)
```

```python
print("Even numbers from 1 to 20:")

for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
total = 0

for i in range(1, 101):
    total += i

print("Sum of numbers from 1 to 100:", total)
```

```python
number = 7

print(f"Multiplication table of {number}:")

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

## 5. Functions

Functions are reusable blocks of code designed to perform specific tasks.

```python
def add_numbers(a, b):
    return a + b


result = add_numbers(10, 5)
print("Result:", result)
```

```python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


print(check_even_odd(10))
print(check_even_odd(7))
```

```python
def factorial(number):
    result = 1

    for i in range(1, number + 1):
        result *= i

    return result


print("Factorial of 5:", factorial(5))
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)


numbers = [10, 20, 30, 40, 50]

print("Average:", calculate_average(numbers))
```

## 6. Object-Oriented Programming (OOP)

This section demonstrates how to create a class and objects in Python.

```python
class Student:

    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Grade:", self.grade)

    def is_passing(self):
        if self.grade >= 50:
            return True
        else:
            return False


student1 = Student("Arzaq", 21, 90)

student1.display_info()

print("Passing:", student1.is_passing())
```

## 7. File Handling

Python provides built-in functions for creating, writing, reading, and updating files.

```python
# Writing to a file

with open("student_info.txt", "w") as file:
    file.write("Name: Arzaq\n")
    file.write("Major: Artificial Intelligence\n")
    file.write("Status: Student\n")

print("File created successfully.")
```

```python
# Reading the file

with open("student_info.txt", "r") as file:
    content = file.read()

print("File content:")
print(content)
```

```python
# Adding more information

with open("student_info.txt", "a") as file:
    file.write("Internship: AI & Machine Learning\n")

print("Information added successfully.")

# Reading updated file

with open("student_info.txt", "r") as file:
    print(file.read())
```

## 8. Exception Handling

Exception handling allows programs to handle errors without crashing.

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

```python
try:
    number = int("abc")
    print(number)

except ValueError:
    print("Error: Invalid number format.")
```

```python
try:
    with open("missing_file.txt", "r") as file:
        content = file.read()

    print(content)

except FileNotFoundError:
    print("Error: File not found.")
```

## 9. Mini Project — Student Grade Management

This mini project combines variables, lists, dictionaries, loops, functions, and conditional statements.

```python
def calculate_average(grades):
    return sum(grades) / len(grades)


students = [
    {
        "name": "Ahmed",
        "grades": [80, 75, 90]
    },
    {
        "name": "Sara",
        "grades": [95, 88, 92]
    },
    {
        "name": "Arzaq",
        "grades": [90, 85, 95]
    }
]

for student in students:
    average = calculate_average(student["grades"])

    print("Student:", student["name"])
    print("Grades:", student["grades"])
    print("Average:", average)

    if average >= 50:
        print("Status: Passed")
    else:
        print("Status: Failed")

    print("-" * 30)
```
