3-preprocessing

```python
benign = df[df["Label"] == "BENIGN"]
attack = df[df["Label"] != "BENIGN"]
```

```python
benign_sample = benign.sample(5000, random_state=42)
attack_sample = attack.sample(5000, random_state=42)
```

```python
df_sample = pd.concat([benign_sample, attack_sample])
```

```python
df_sample["Label"].value_counts()
```

```python
import numpy as np

df_sample.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
from sklearn.impute import SimpleImputer

num_cols = df_sample.select_dtypes(include=["int64","float64"]).columns

imputer = SimpleImputer(strategy="median")

df_sample[num_cols] = imputer.fit_transform(df_sample[num_cols])
```

```python
df_sample["Label"] = df_sample["Label"].apply(
    lambda x: 0 if x == "BENIGN" else 1
)
```

```python
df_sample.isnull().sum().sort_values(ascending=False).head(20)
```

```python
df_sample
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x="Label", data=df_sample)

plt.title("Class Distribution")
plt.show()
```

```python
# حفظ الداتا بعد preprocessing
df_sample.to_csv("processed_dataset.csv", index=False)
```
